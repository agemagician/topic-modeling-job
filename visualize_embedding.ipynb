```python
#read the previously processed data
import pandas as pd
import json
import umap
from sklearn.manifold import TSNE
import matplotlib.pyplot as plt
```

```python
data = pd.read_pickle("output/preprocessed_data/use_this_data.pkl")
sentences = list(data['sentences'])
token_lists = list(data['token_list'])
```

```python
saved_embedding_dir = 'output/preprocessed_data/bert_embeddings.json'
```

```python
with open(saved_embedding_dir) as json_file:
    sentence_embedding = json.load(json_file)
```

```python
sentence_embedding
```

```python
reducer = umap.UMAP()
```

```python
vec_umap = reducer.fit_transform(sentence_embedding)
```

```python
vec_umap
```

```python
plt.plot(vec_umap[:, 0], vec_umap[:, 1], '.', alpha=0.5,
                 label='some dummy label')
plt.legend(loc = 'best')
plt.grid(color ='grey', linestyle='-',linewidth = 0.25)
```

```python
#creating files to visualize the sentence embeddings in TensorBoard
import io

out_v = io.open('vecs.tsv', 'w', encoding='utf-8')
out_m = io.open('meta.tsv', 'w', encoding='utf-8')

for num, word in enumerate(sentences):
    vec = sentence_embedding[num] 
    out_m.write(word + "\n")
    out_v.write('\t'.join([str(x) for x in vec]) + "\n")
out_v.close()
out_m.close()
```
